```python
import numpy as np
```

```python
arr_zeros = np.zeros(10)
print(arr_zeros)
```

```python
arr_ones = np.ones(10)
print(arr_ones)
```

```python
arr_fives = np.ones(10) * 5
print(arr_fives)
```

```python
arr_10_50 = np.arange(10, 51)
print(arr_10_50)
```

```python
arr_even = np.arange(10, 51, 2)
print(arr_even)
```

```python
matrix_3x3 = np.arange(9).reshape(3, 3)
print(matrix_3x3)
```

```python
identity_matrix = np.eye(3)
print(identity_matrix)
```

```python
rand_num = np.random.rand()
print(rand_num)
```

```python
rand_normal = np.random.randn(25)
print(rand_normal)
```
