```python
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

# Plotting
plt.plot(x, y, 'r')   # 'r' means red color

# Labels and Title
plt.xlabel("X Axis Title Here")
plt.ylabel("Y Axis Title Here")
plt.title("String Title Here")

# Display the graph
plt.show()
```
