```python
import pandas as pd

# Sample data for each column
roll_numbers = [101, 102, 103, 104]
student_names = ['Alice', 'Bob', 'Charlie', 'David']
sections = ['A', 'B', 'A', 'C']
marks1 = [85, 90, 78, 92]
marks2 = [92, 88, 80, 85]
marks3 = [78, 95, 88, 90]

# Create a dictionary with column names as keys and data lists as values
data = {
    'Roll number': roll_numbers,
    'name': student_names,
    'Section': sections,
    'Marks1': marks1,
    'Marks2': marks2,
    'Marks3': marks3
}

# Create the DataFrame
df = pd.DataFrame(data, index=range(4))   # default index

# Change index from 0..3 to row A, row B, row C, row D
ind = ['row A', 'row B', 'row C', 'row D']
df.index = ind

# Print the DataFrame
print(df)
```

```python
import pandas as pd

internal_marks = pd.Series(
    [35, 42, 28, 40, 45, 38, 30, 47, 41, 33],
    index=[101,102,103,104,105,106,107,108,109,110]
)

endsem_marks = pd.Series(
    [40, 44, 32, 45, 48, 41, 35, 46, 43, 38],
    index=[101,102,103,104,105,106,107,108,109,110]
)

print(internal_marks)
print(endsem_marks)
```

```python
print("Max Internal Marks Roll No:", internal_marks.idxmax())
print("Max End-Sem Marks Roll No:", endsem_marks.idxmax())
```

```python
total_marks = internal_marks + endsem_marks
print(total_marks)
```
