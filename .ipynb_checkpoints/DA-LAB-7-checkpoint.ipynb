```python
import pandas as pd
import numpy as np

# Create dataframe
df = pd.DataFrame({
    'A':[1,2,np.nan],
    'B':[5,np.nan,np.nan],
    'C':[1,2,3]
})

print("Original DataFrame:")
print(df)

# Drop rows containing NaN
print("\nDrop rows with NaN:")
print(df.dropna())

# Drop columns containing NaN
print("\nDrop columns with NaN:")
print(df.dropna(axis=1))

# Remove rows with at least 2 non-NaN values
print("\nThreshold drop:")
print(df.dropna(thresh=2))

# Fill NaN with constant value
print("\nFill NaN with VAL:")
print(df.fillna(value='VAL'))

# Fill NaN in column A with mean
print("\nFill NaN in A with mean:")
print(df['A'].fillna(value=df['A'].mean()))
```

```python
import pandas as pd

data = {
    'Company':['GOOGLE','GOOGLE','MICROSOFT','MICROSOFT','FB','FB'],
    'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
    'Sales':[200,120,340,124,243,350]
}

df = pd.DataFrame(data)

print(df)

# Group by Company
by_comp = df.groupby("Company")

# Mean sales
print("\nMean Sales:")
print(by_comp['Sales'].mean())

# Standard deviation
print("\nStandard Deviation:")
print(by_comp['Sales'].std())

# Minimum values
print("\nMinimum values:")
print(by_comp.min())

# Describe statistics
print("\nDescribe:")
print(by_comp.describe())

# Transpose describe
print("\nTranspose describe:")
print(by_comp.describe().transpose())

print("\nGOOGLE stats:")
print(by_comp.describe().transpose()['GOOGLE'])
```

```python
df1 = pd.DataFrame({
    'A':['A0','A1','A2','A3'],
    'B':['B0','B1','B2','B3'],
    'C':['C0','C1','C2','C3'],
    'D':['D0','D1','D2','D3']
}, index=[0,1,2,3])

df2 = pd.DataFrame({
    'A':['A4','A5','A6','A7'],
    'B':['B4','B5','B6','B7'],
    'C':['C4','C5','C6','C7'],
    'D':['D4','D5','D6','D7']
}, index=[4,5,6,7])

df3 = pd.DataFrame({
    'A':['A8','A9','A10','A11'],
    'B':['B8','B9','B10','B11'],
    'C':['C8','C9','C10','C11'],
    'D':['D8','D9','D10','D11']
}, index=[8,9,10,11])

# Row concatenation
print(pd.concat([df1,df2,df3]))

# Column concatenation
print(pd.concat([df1,df2,df3], axis=1))
```

```python
left = pd.DataFrame({
    'key':['K0','K1','K2','K3'],
    'A':['A0','A1','A2','A3'],
    'B':['B0','B1','B2','B3']
})

right = pd.DataFrame({
    'key':['K0','K1','K2','K3'],
    'C':['C0','C1','C2','C3'],
    'D':['D0','D1','D2','D3']
})

print(pd.merge(left, right, how='inner', on='key'))
```

```python
left = pd.DataFrame({
    'key1':['K0','K0','K1','K2'],
    'key2':['K0','K1','K0','K1'],
    'A':['A0','A1','A2','A3'],
    'B':['B0','B1','B2','B3']
})

right = pd.DataFrame({
    'key1':['K0','K1','K1','K2'],
    'key2':['K0','K0','K0','K0'],
    'C':['C0','C1','C2','C3'],
    'D':['D0','D1','D2','D3']
})

print(pd.merge(left, right, on=['key1','key2']))
print(pd.merge(left, right, how='outer', on=['key1','key2']))
print(pd.merge(left, right, how='right', on=['key1','key2']))
print(pd.merge(left, right, how='left', on=['key1','key2']))
```

```python
left = pd.DataFrame({
    'A':['A0','A1','A2'],
    'B':['B0','B1','B2']
}, index=['K0','K1','K2'])

right = pd.DataFrame({
    'C':['C0','C2','C3'],
    'D':['D0','D2','D3']
}, index=['K0','K2','K3'])

# Default join (left)
print(left.join(right))

# Outer join
print(left.join(right, how='outer'))
```
